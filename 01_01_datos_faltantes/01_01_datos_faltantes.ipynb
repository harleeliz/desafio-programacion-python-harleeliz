```python
#Mapeado de las columnas

COLUMNS=['Nombre', 'Sexo', 'Estado_Civil', 'Edad', 'Experiencia',
       'Salario', 'Gerencia', 'Departamento']
```

## **1. Limpieza de Columnas** <br>
>Nota. Toda la columna de Apellidos tiene exactamente 251 campos en blanco - por lo que elimino columna

```python
df_NoLastName=df.drop(axis=1,columns=['Apellido'])
print(df_NoLastName.shape)
print(df_NoLastName.columns)
```

## **2. Limpieza de Valores en con espacios y pasandolos a NaN** <br>
Casos tipo ' ' o "  "

```python
#Pasando vacio a NaN
df_NaN=df_NoLastName.replace(r'^\s*$', pd.NA, regex=True)
print(df_NaN.isna().sum())
```

## **Imputación de datos- Relleno de edad** <br>
En este apartado modifico las edades faltantes para no eliminarlas completamente

```python
dummyAge=round(df['Edad'].mean(),0)
print(dummyAge)

df_NaN['Edad']=df_NaN['Edad'].fillna(dummyAge)
```

## **3.Limpieza de Valores NaN** <br>
Es esta forma borro todos las filas que contengan un NaN value

```python
#Nota, Declarar las columnas me permite por ejemplo eliminar de ciertas columnas y no borrar de todas.
df_clean = df_NaN.dropna(subset=COLUMNS)
print(df_clean.isna().sum())
print(df_clean.head())
```

```python
#Dataset check
print(df_clean.shape)
print(df_clean.isnull().sum())
print(df_clean.isna().sum())
```

## **4. Limpieza de filas duplicadas**

```python
#Identificacion de duplicados
duplicates = df_clean[df.duplicated(keep=False)]
print(duplicates)
print(df_clean.shape)
```

```python
df_NoDupl=df_clean.drop_duplicates(subset=COLUMNS)
print(df_NoDupl.shape) #Se eliminaron 2 filas
df_NoDupl.head(5)
```

```python
df_clean_sorted=df_NoDupl.sort_values(by=COLUMNS[0],ignore_index=True)
```

```python
#New Dataset - creando nuevo dataset
file_name="new_empleados.csv"
df_clean_sorted.to_csv(file_name,sep=";",index=False,encoding="utf-8-sig")
```
